```python
# Sum of digits in a number
number = int(input("Enter number: "))

total = 0

while number > 0:
    digit = number % 10
    total = total + digit
    number = number // 10

print("Sum of digits:", total)    
```

```python
 # Reverse a number
number = int(input("Enter number: "))

reverse = 0

while number > 0:
    digit = number % 10
    number = number // 10
    reverse = reverse * 10 + digit

print("Reverse: ", reverse)
```

```python
number = int(input("Enter number:"))

original = number
reverse = 0

while number > 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number = number // 10

if original == reverse:
    print("pailndrome")
else:
    print("Not pailndrome")
```

```python
# Check if a number is prime
number = int(input("Enter number: "))

count = 0

for i in range(1, number + 1):
    if number % 1 == 0:
        count = count + 1

if count == 2:
    print("Prime number")
else:
    print("Not a prime number")
```

```python
# Print all prime numbers between 2 and 100
for number in range(2, 101):

    count = 0

    for i in range(1, number + 1):
        if number % i == 0:
            count = count + 1

    if count == 2:
        print(number)
```
