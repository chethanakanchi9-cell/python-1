```python
#list in python
#list is an ordered and changeable collection that can store multiple values 
marks = [80, 90, 75, 85]

print(marks)
```

```python
#accessing elements in a list
marks = [80, 90, 75, 85]

print(marks[0])
print(marks[1])
print(marks[2])
```

```python
#change elements in a list
marks = [80, 90, 75]

marks[1] = 95

print(marks)
```

```python
#change elements in a list
marks = [79, 64, 92, 74, 96, 77, 94]

marks[1] = 65

marks[3] = 75

marks[5] = 79

print(marks)
```

```python
#add elements to a list
marks = [80, 90, 75]

marks.append(85)

print(marks)
```

```python
#add elements to a list
marks = [25, 45]

marks.append(90)

marks.append(75)

print(marks)
```

```python
#remove elements from a list
marks = [80, 90, 75]

marks.remove(90)

print(marks)
```
