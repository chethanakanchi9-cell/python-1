```python

#dictionaries in python
#Dictionary is a collection od key-value pairs that is unordered and mutable
student = {
    "name": "Chethana",
    "age": 18,
    "course": "Python"
}
print(student)
```

```python
#access elements in dic
print(student["name"])
print(student["age"])
print(student["course"])
```

```python
#add new data to a  dictionary
student["city"] = "Tirupati"

print(student)
```

```python

student = {
    "name": "Chethana",
    "age": 18,
    "course": "Python"
}

print(student.keys())
# keys() returns all the keys in the dictionary

print(student.values())
# values() returns all the values in the dictionary

print(student.items())
# items() returns all the key-value pairs

print(student.get("name"))
# get() returns the value of the specified key

student.update({"age": 19})
# update() updates the value of the specified key

print(student)


student.pop("age")
# pop() removes the specified key and its value

print(student)
```

```python
#popitem() removes the last inserted key-value pair
student = {
     "name" : "Chethana",
      "age": 18,
      "course" : "python"
}

student.popitem()

print(student)
```

```python
student = {
    "name": "Chethana"
}

student.setdefault("age", 18)

print(student)
```
